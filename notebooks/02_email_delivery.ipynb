# Notebook 2: Email Sender Service

This notebook demonstrates how to send an HTML email using Python.

The email service is implemented using the Brevo Transactional Email API. This is used instead of Gmail SMTP because Gmail App Passwords may not be available for all accounts, and API-based transactional email services are more suitable for production-like workflows.

The notebook accepts a receiver email address and sends a formatted HTML email.

## Receiver Email Input

Provide the receiver email address below.

```python
TO_EMAIL = "arghya79@gmail.com"

EMAIL_SUBJECT = "Test Email - Open Weather Challenge"
```

```python
sample_html_body = f"""
<html>
  <body style="margin:0; padding:0; background-color:#f4f7fb; font-family:Arial, sans-serif;">
    <div style="max-width:720px; margin:30px auto; background-color:#ffffff; border-radius:14px; padding:28px; border:1px solid #e5e7eb;">

      <h2 style="color:#1f4e79; margin-bottom:8px;">
        Open Weather Challenge - Email Service Test
      </h2>

      <p style="color:#374151; font-size:15px;">
        Hello,
      </p>

      <p style="color:#374151; font-size:15px; line-height:1.6;">
        This is a test email sent from Python using the Brevo Transactional Email API.
        The objective of this notebook is to validate that the email-sending function works correctly.
      </p>

      <table style="border-collapse:collapse; width:100%; margin-top:18px; font-size:14px;">
        <tr>
          <td style="padding:12px; border:1px solid #e5e7eb; background-color:#f9fafb;"><b>Status</b></td>
          <td style="padding:12px; border:1px solid #e5e7eb;">Email service working successfully</td>
        </tr>
        <tr>
          <td style="padding:12px; border:1px solid #e5e7eb; background-color:#f9fafb;"><b>Generated At</b></td>
          <td style="padding:12px; border:1px solid #e5e7eb;">{datetime.now().strftime("%Y-%m-%d %H:%M:%S")}</td>
        </tr>
        <tr>
          <td style="padding:12px; border:1px solid #e5e7eb; background-color:#f9fafb;"><b>Project</b></td>
          <td style="padding:12px; border:1px solid #e5e7eb;">Daily Weather Forecast Email Service</td>
        </tr>
      </table>

      <p style="margin-top:24px; color:#374151; font-size:15px;">
        Regards,<br>
        <b>Weather Forecast Email Service</b>
      </p>

    </div>
  </body>
</html>
"""
```

```python
def send_email_with_brevo(
    to_email,
    subject,
    html_body,
    sender_email,
    sender_name,
    brevo_api_key
):
    url = "https://api.brevo.com/v3/smtp/email"

    headers = {
        "accept": "application/json",
        "api-key": brevo_api_key,
        "content-type": "application/json"
    }

    payload = {
        "sender": {
            "name": sender_name,
            "email": sender_email
        },
        "to": [
            {
                "email": to_email
            }
        ],
        "subject": subject,
        "htmlContent": html_body,
        "textContent": "This is a test email from the Open Weather Challenge email service."
    }

    response = requests.post(url, headers=headers, json=payload, timeout=30)

    if response.status_code not in [200, 201, 202]:
        raise Exception(
            f"Email sending failed. Status Code: {response.status_code}, Response: {response.text}"
        )

    return response.json()
```

## Test Mail Through Brevo API Transactional Service

### Reason for Choosing Brevo

Brevo was selected for the notebook-based demonstration because it provides a simple transactional email API that can be called from Python using HTTP requests.

Gmail SMTP was avoided because Gmail App Passwords may not be available for every account and are not ideal for production workflows.

For this assignment, Brevo provides a simple way to demonstrate the email functionality securely from Google Colab without hardcoding credentials.

```python
email_response = send_email_with_brevo(
    to_email=TO_EMAIL,
    subject=EMAIL_SUBJECT,
    html_body=sample_html_body,
    sender_email=SENDER_EMAIL,
    sender_name=SENDER_NAME,
    brevo_api_key=BREVO_API_KEY
)

print("Email sent successfully.")
print(email_response)
```
